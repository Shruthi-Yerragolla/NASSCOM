## 1. What is pandas?

pandas is a Python package that provides fast, flexible, and expressive data structures for working with 'relational' or 'labeled' data. It is built on top of NumPy, was developed by **Wes McKinney in 2008**, and the name comes from **"Panel Data"**.

It provides **three** core data structures:

| Data Structure | Dimensions | Description |
|---|---|---|
| **Series** | 1 | 1D labeled, homogeneous array, size-immutable |
| **DataFrame** | 2 | 2D labeled, size-mutable, heterogeneously-typed columns |
| **Panel** | 3 | 3D labeled, size-mutable (legacy — rarely used today) |

## 2. Creating a Series

```python
# -----------------------------------------------------------
# 🔹 2A. DEMO — Empty Series
# -----------------------------------------------------------
s = pd.Series(dtype='float64')
print(s)
```

```python
# -----------------------------------------------------------
# 🔹 2B. DEMO — Series from an ndarray (default integer index)
# -----------------------------------------------------------
info = np.array(['P', 'a', 'n', 'd', 'a', 's'])
s = pd.Series(info)
print(s)
```

```python
# -----------------------------------------------------------
# 🔹 2D. DEMO — Series from a scalar (value repeated to match index)
# -----------------------------------------------------------
s = pd.Series(5, index=[0, 1, 2, 3])
print(s)
```

## 3. Accessing Data in a Series

```python
# -----------------------------------------------------------
# 🔹 3A. DEMO — access by position and by label
# -----------------------------------------------------------
s = pd.Series([1, 2, 3, 4, 5], index=['a', 'b', 'c', 'd', 'e'])

print('First element (position 0):', s.iloc[0])
print('First three elements:')
print(s.iloc[:3])

print('\nElement at label "a":', s.loc['a'])
print('Elements at labels a, c, d:')
print(s.loc[['a', 'c', 'd']])
```

## 4. Creating a DataFrame

```python
# -----------------------------------------------------------
# 🔹 4A. DEMO — DataFrame from a list of lists
# -----------------------------------------------------------
data = [['Alex', 10], ['Bob', 12], ['Clarke', 13]]
df = pd.DataFrame(data, columns=['Name', 'Age'])
print(df)
```

```python
# -----------------------------------------------------------
# 🔹 4B. DEMO — DataFrame from a dict of lists
# -----------------------------------------------------------
data = {'Name': ['Tom', 'Jack', 'Steve', 'Ricky'], 'Age': [28, 34, 29, 42]}
df = pd.DataFrame(data, index=['rank1', 'rank2', 'rank3', 'rank4'])
print(df)
```

```python
# -----------------------------------------------------------
# 🔹 4C. DEMO — DataFrame from a list of dicts (missing keys -> NaN)
# -----------------------------------------------------------
data = [{'a': 1, 'b': 2}, {'a': 5, 'b': 10, 'c': 20}]
df = pd.DataFrame(data)
print(df)
```

```python
# -----------------------------------------------------------
# 🔹 4D. DEMO — DataFrame from a dict of Series (union of indexes)
# -----------------------------------------------------------
d = {
    'one': pd.Series([1, 2, 3], index=['a', 'b', 'c']),
    'two': pd.Series([1, 2, 3, 4], index=['a', 'b', 'c', 'd']),
}
df = pd.DataFrame(d)
print(df)
```

## 5. Column Selection, Addition & Deletion

```python
# -----------------------------------------------------------
# 🔹 5A. DEMO — select, add, and delete columns
# -----------------------------------------------------------
d = {
    'one': pd.Series([1, 2, 3], index=['a', 'b', 'c']),
    'two': pd.Series([1, 2, 3, 4], index=['a', 'b', 'c', 'd']),
}
df = pd.DataFrame(d)

print("Select column 'one':")
print(df['one'])

print("\nAdd column 'three' from a Series:")
df['three'] = pd.Series([10, 20, 30], index=['a', 'b', 'c'])
print(df)

print("\nAdd column 'four' from existing columns:")
df['four'] = df['one'] + df['three']
print(df)

print("\nDelete column 'one' with del:")
del df['one']
print(df)

print("\nDelete column 'two' with pop:")
df.pop('two')
print(df)
```

## 6. Row Selection, Addition & Deletion

```python
# -----------------------------------------------------------
# 🔹 6A. DEMO — select rows by label (loc), by position (iloc), and by slice
# -----------------------------------------------------------
d = {
    'one': pd.Series([1, 2, 3], index=['a', 'b', 'c']),
    'two': pd.Series([1, 2, 3, 4], index=['a', 'b', 'c', 'd']),
}
df = pd.DataFrame(d)

print("Row 'b' by label (.loc):")
print(df.loc['b'])

print("\nRow at position 2 (.iloc):")
print(df.iloc[2])

print("\nSlice rows 2 to 4:")
print(df[2:4])
```

```python
# -----------------------------------------------------------
# 🔹 6B. DEMO — add and drop rows
# -----------------------------------------------------------
# Note: DataFrame.append() was removed in modern pandas — pd.concat() replaces it.
df1 = pd.DataFrame([[1, 2], [3, 4]], columns=['a', 'b'])
df2 = pd.DataFrame([[5, 6], [7, 8]], columns=['a', 'b'])

combined = pd.concat([df1, df2], ignore_index=False)
print('Combined (duplicate labels kept):')
print(combined)

print("\nDrop rows with label 0:")
dropped = combined.drop(0)
print(dropped)
```
